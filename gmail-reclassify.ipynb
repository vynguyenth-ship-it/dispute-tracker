```python
import os
from google.auth.transport.requests import Request
from google.oauth2.credentials import Credentials
from google_auth_oauthlib.flow import InstalledAppFlow
from googleapiclient.discovery import build

SCOPES = ["https://www.googleapis.com/auth/gmail.modify"]

def get_gmail_service():
    creds = None
    if os.path.exists("token.json"):
        creds = Credentials.from_authorized_user_file("token.json", SCOPES)

    if not creds or not creds.valid:
        if creds and creds.expired and creds.refresh_token:
            creds.refresh(Request())
        else:
            flow = InstalledAppFlow.from_client_secrets_file("credentials.json", SCOPES)
            creds = flow.run_local_server(port=0)
        with open("token.json", "w") as token:
            token.write(creds.to_json())

    return build("gmail", "v1", credentials=creds)
```

```python
import os
from google.auth.transport.requests import Request
from google.oauth2.credentials import Credentials
from google_auth_oauthlib.flow import InstalledAppFlow
from googleapiclient.discovery import build

SCOPES = ["https://www.googleapis.com/auth/gmail.modify"]

def get_gmail_service():
    creds = None
    
    # 1️⃣ Check for existing user token (created after previous login)
    if os.path.exists("token.json"):
        creds = Credentials.from_authorized_user_file("token.json", SCOPES)

    # 2️⃣ If no valid token, do the login flow
    if not creds or not creds.valid:
        if creds and creds.expired and creds.refresh_token:
            creds.refresh(Request())
        else:
            # This uses credentials.json (app identity) to start login
            flow = InstalledAppFlow.from_client_secrets_file("credentials.json", SCOPES)
            creds = flow.run_local_server(port=0)
        
        # 3️⃣ Save the user token for next time
        with open("token.json", "w") as token:
            token.write(creds.to_json())

    return build("gmail", "v1", credentials=creds)
```

```python
import os
import base64
from google.auth.transport.requests import Request
from google.oauth2.credentials import Credentials
from google_auth_oauthlib.flow import InstalledAppFlow
from googleapiclient.discovery import build

# Make sure we're in the right folder
os.chdir(r"C:\Users\vy.nguyenth\gmail-classifer")
print("📂 Working in:", os.getcwd())
print("📄 Files:", os.listdir("."))

SCOPES = ["https://www.googleapis.com/auth/gmail.modify"]

def get_gmail_service():
    creds = None
    
    if os.path.exists("token.json"):
        creds = Credentials.from_authorized_user_file("token.json", SCOPES)

    if not creds or not creds.valid:
        if creds and creds.expired and creds.refresh_token:
            creds.refresh(Request())
        else:
            flow = InstalledAppFlow.from_client_secrets_file("credentials.json", SCOPES)
            creds = flow.run_local_server(port=0)
        
        with open("token.json", "w") as token:
            token.write(creds.to_json())

    return build("gmail", "v1", credentials=creds)

print("\n✅ Auth function ready")
```

```python
print("🔌 Connecting to Gmail API...")
service = get_gmail_service()

profile = service.users().getProfile(userId="me").execute()
print(f"\n✅ Connected as: {profile['emailAddress']}")
print(f"📬 Total messages: {profile['messagesTotal']}")
```

```python
print("📨 Recent inbox emails:\n")
results = service.users().messages().list(userId="me", q="in:inbox", maxResults=5).execute()

for m in results.get("messages", []):
    msg = service.users().messages().get(userId="me", id=m["id"], format="metadata").execute()
    headers = {h["name"]: h["value"] for h in msg["payload"]["headers"]}
    print(f"  - {headers.get('Subject', '(no subject)')[:70]}")
```

```python
CATEGORIES = ["invoice", "support", "newsletter", "personal", "other"]

def get_email_body(payload):
    if "parts" in payload:
        for part in payload["parts"]:
            if part["mimeType"] == "text/plain" and "data" in part.get("body", {}):
                return base64.urlsafe_b64decode(part["body"]["data"]).decode("utf-8", errors="ignore")
    elif "data" in payload.get("body", {}):
        return base64.urlsafe_b64decode(payload["body"]["data"]).decode("utf-8", errors="ignore")
    return ""

def classify_email(subject, body):
    text = (subject + " " + body).lower()
    if any(k in text for k in ["hóa đơn", "điều chỉnh", "chiết khấu"]):
        return "invoice"
    if any(k in text for k in ["unsubscribe", "newsletter", "weekly digest"]):
        return "newsletter"
    if any(k in text for k in ["help", "ticket", "support", "issue"]):
        return "support"
    return "other"

def get_or_create_label(service, name):
    labels = service.users().labels().list(userId="me").execute().get("labels", [])
    for lbl in labels:
        if lbl["name"].lower() == name.lower():
            return lbl["id"]
    new_lbl = service.users().labels().create(
        userId="me",
        body={"name": name, "labelListVisibility": "labelShow", "messageListVisibility": "show"}
    ).execute()
    print(f"✨ Created label: {name}")
    return new_lbl["id"]

def reclassify_inbox(max_emails=5):
    # 👇 No more "auto/" prefix
    label_ids = {cat: get_or_create_label(service, cat) for cat in CATEGORIES}
    processed_label = get_or_create_label(service, "processed")

    # 👇 Update query to match new label name
    results = service.users().messages().list(
        userId="me",
        q="in:inbox -label:processed",
        maxResults=max_emails
    ).execute()

    messages = results.get("messages", [])
    print(f"\n🔍 Found {len(messages)} emails to classify.\n")

    for i, m in enumerate(messages, 1):
        msg = service.users().messages().get(userId="me", id=m["id"], format="full").execute()
        headers = {h["name"]: h["value"] for h in msg["payload"]["headers"]}
        subject = headers.get("Subject", "(no subject)")
        body = get_email_body(msg["payload"])[:1500]

        category = classify_email(subject, body)
        print(f"[{i}/{len(messages)}] [{category:10}] {subject[:60]}")

        service.users().messages().modify(
            userId="me",
            id=m["id"],
            body={"addLabelIds": [label_ids[category], processed_label]}
        ).execute()
    
    print(f"\n✅ Done! Labeled {len(messages)} emails.")
    print("   Refresh Gmail to see the labels in the left sidebar.")

reclassify_inbox(max_emails=5)
```

```python
labels = service.users().labels().list(userId="me").execute().get("labels", [])

deleted = 0
for lbl in labels:
    if lbl["name"].startswith("auto/"):
        service.users().labels().delete(userId="me", id=lbl["id"]).execute()
        print(f"🗑️ Deleted: {lbl['name']}")
        deleted += 1

print(f"\n✅ Removed {deleted} old auto/* labels")
```
